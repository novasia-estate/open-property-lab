# Floor-plan Region Topology Lab

Review overlaps, gaps and interfaces between explicitly supplied orthogonal regions. Five synthetic examples are shown below. Areas use caller-supplied drawing units, with exact rational arithmetic. No image tracing, measured-area, structural or legal conclusion is made.

Prepared by [NovAsia Estate](https://novasia.estate/). [Source, tests and input schema](https://github.com/novasia-estate/open-property-lab/tree/main/region-topology-lab). MIT.


```python
import json
from pathlib import Path
from html import escape
from IPython.display import HTML, display
import topology
from topology import analyse, svg
root = Path(topology.__file__).parent
examples = json.loads((root / 'examples.json').read_text())
def show_case(name, document):
    report = analyse(document)
    rows = [('Envelope area', report['envelope_area']), ('Uncovered area', report['uncovered_area']), ('Multiply covered area', report['multiply_covered_area'])]
    text = '<h3>' + escape(name.replace('_', ' ').title()) + '</h3>' + svg(document)
    text += '<table><tr><th>Measure</th><th>Exact drawing units squared</th></tr>'
    text += ''.join('<tr><td>'+escape(k)+'</td><td>'+escape(v)+'</td></tr>' for k,v in rows)+'</table>'
    text += '<p>Regions: '+escape(', '.join(r['id'] for r in document['regions']))+'</p>'
    text += '<p>Exclusive interfaces: '+escape(str(report['exclusive_interfaces']))+'</p>'
    text += '<p>Outside envelope: '+escape(str(report['outside_envelope_area']))+'</p>'
    return text
```

## Compare five cases

A gap may be an intentional void. An exclusive interface is not evidence of a physical wall or door. Overlap area counts each multiply occupied cell once, including triple overlaps. SVG diagrams illustrate the supplied annotation; the exact report is the numerical result.


```python
display(HTML(''.join(show_case(name, document) for name, document in examples.items())))
```
