## 1. Coverage comes before comparison

A latest value is not necessarily a current value. The table identifies the last reported year and missing requested years. Do not put different countries’ latest observations into a chart labelled with one common year.


```python
summary=[]
for country in countries:
    available=[r for r in rows if r['country']==country and r['international_arrivals']]
    latest=max(available,key=lambda r:int(r['year']))
    missing=[r['year'] for r in rows if r['country']==country and not r['international_arrivals']]
    summary.append([country,str(len(available)),latest['year'],f"{float(latest['international_arrivals']):,.0f}",', '.join(missing) or 'none'])
table='<table><thead><tr>'+''.join('<th>'+c+'</th>' for c in ['Country','Reported years','Latest year','Latest arrivals','Missing2000–2025'])+'</tr></thead><tbody>'
table+=''.join('<tr>'+''.join('<td>'+html.escape(v)+'</td>' for v in row)+'</tr>' for row in summary)+'</tbody></table>'
display(HTML(table))
```

## 2. Recovery uses the same explicit baseline

Each country’s 2019 observation is 100. A value of 80 means reported arrivals equal 80% of that country’s 2019 count. It does not mean 80% hotel occupancy. Missing years break the plotted line rather than being interpolated.


```python
colors={'Cambodia':'#D47A46','Indonesia':'#326B83','Thailand':'#247461'}
values=[float(r['recovery_2019_100']) for r in rows if r['recovery_2019_100']]
top=max(120,max(values)*1.05)
def x(year):return 70+(int(year)-2000)*29
def y(value):return 330-float(value)/top*260
svg=['<svg xmlns="http://www.w3.org/2000/svg" width="900" height="410" viewBox="0 0 900 410"><rect width="900" height="410" fill="#f6f8f5"/><text x="70" y="26" font-size="18">Annual international arrivals:2019=100</text>']
for tick in [0,50,100]:
    yy=y(tick);svg.append(f'<line x1="70" y1="{yy}" x2="800" y2="{yy}" stroke="#b8c9c2"/><text x="25" y="{yy+4}">{tick}</text>')
for country in countries:
    previous=None
    for r in sorted([r for r in rows if r['country']==country],key=lambda r:int(r['year'])):
        if not r['recovery_2019_100']:previous=None;continue
        point=(x(r['year']),y(r['recovery_2019_100']))
        if previous:svg.append(f'<line x1="{previous[0]}" y1="{previous[1]}" x2="{point[0]}" y2="{point[1]}" stroke="{colors[country]}" stroke-width="2"/>')
        svg.append(f'<circle cx="{point[0]}" cy="{point[1]}" r="2.5" fill="{colors[country]}"><title>{country} {r["year"]}: {r["recovery_2019_100"]}</title></circle>');previous=point
for year in [2000,2005,2010,2015,2020,2025]:svg.append(f'<text x="{x(year)-15}" y="360">{year}</text>')
for i,country in enumerate(countries):svg.append(f'<text x="{70+i*230}" y="392" fill="{colors[country]}">{country}</text>')
svg.append('</svg>');display(SVG(''.join(svg)))
```

## 3. Reproducible conclusions and boundaries

- Check the coverage table before choosing a reporting year.
- Compare against the same explicit baseline; retain raw counts beside indexed values.
- Do not average countries’ recovery indices as a regional growth rate: their original arrival counts and definitions differ.
- National annual tourism counts do not establish neighbourhood demand, seasonality, nightly rates or a rental forecast.
- Refreshing the upstream series can revise historical observations. Keep this snapshot and cite the retrieval date in `provenance.json`.
